## Module 2: Control Flow

```python
# ─────────────────────────────────────────
# 2.1 Conditionals — Signal Classification
# ─────────────────────────────────────────

def classify_signal(rsi: float) -> str:
    """Classify RSI into trading signals."""
    if rsi >= 70:
        return "OVERBOUGHT — Consider selling"
    elif rsi >= 60:
        return "BULLISH — Hold / add on dips"
    elif rsi >= 40:
        return "NEUTRAL — No clear signal"
    elif rsi >= 30:
        return "BEARISH — Reduce exposure"
    else:
        return "OVERSOLD — Consider buying"


for rsi_val in [15, 32, 55, 65, 78]:
    print(f"RSI {rsi_val:>3}: {classify_signal(rsi_val)}")


# Ternary (conditional expression)
price_now = 185.0
stop_loss = 175.0
action = "HOLD" if price_now > stop_loss else "STOP-LOSS TRIGGERED"
print(f"\nAction: {action}")


# Match statement (Python 3.10+) — order routing
order_type = "market"
match order_type:
    case "market":
        print("\nExecute immediately at best available price")
    case "limit":
        print("Execute only at specified price or better")
    case "stop":
        print("Trigger when price reaches stop level")
    case _:
        print("Unknown order type")


# ─────────────────────────────────────────
# 2.2 Loops — Backtesting Logic
# ─────────────────────────────────────────

closing_prices = [100.0, 102.5, 101.8, 105.3, 103.0, 107.2, 110.0, 108.5]

# FOR loop — compute daily returns
daily_returns = []
for i in range(1, len(closing_prices)):
    ret = (closing_prices[i] - closing_prices[i-1]) / closing_prices[i-1]
    daily_returns.append(ret)

print("Daily Returns:")
for i, r in enumerate(daily_returns, start=1):
    direction = "▲" if r > 0 else "▼"
    print(f"  Day {i}: {direction} {r*100:+.2f}%")


# WHILE loop — simulate until drawdown limit hit
print("\nDrawdown Monitor:")
capital = 100_000
max_drawdown_limit = 0.10
peak = capital
day = 0
simulated_losses = [0, -200, -500, -1200, -800, -9500, -1000]

while day < len(simulated_losses):
    capital += simulated_losses[day]
    peak = max(peak, capital)
    drawdown = (peak - capital) / peak
    print(f"  Day {day+1}: Capital=${capital:,.0f} | Drawdown={drawdown:.2%}")
    if drawdown >= max_drawdown_limit:
        print("  ⚠️  MAX DRAWDOWN BREACHED — Stopping trading!")
        break
    day += 1


# Loop control: continue, break, else
print("\nFiltering valid prices (skip zeros):")
raw_prices = [102.5, 0, 103.1, 0, 104.5, 105.0]

for p in raw_prices:
    if p == 0:
        continue
    print(f"  Valid price: {p}")
else:
    print("  All prices processed.")
```

## Module 4: Functions

### First-Class Functions, Lambdas, Closures

```python
# ─────────────────────────────────────────
# 4.1 Function Definitions & Arguments
# ─────────────────────────────────────────

import math
import random

def sharpe_ratio(
    returns: list,
    risk_free_rate: float = 0.05,
    periods: int = 252
) -> float:
    """
    Compute the annualized Sharpe Ratio.
    """
    n = len(returns)
    mean_ret = sum(returns) / n
    variance = sum((r - mean_ret) ** 2 for r in returns) / (n - 1)
    std_dev  = math.sqrt(variance)

    rf_daily = risk_free_rate / periods
    excess   = mean_ret - rf_daily

    return (excess / std_dev) * math.sqrt(periods)


random.seed(42)
sim_returns = [random.gauss(0.0005, 0.01) for _ in range(252)]

sr = sharpe_ratio(sim_returns)
print(f"Sharpe Ratio: {sr:.4f}")
print(f"Sharpe (2% rf): {sharpe_ratio(sim_returns, risk_free_rate=0.02):.4f}")


def weighted_portfolio_return(*args, **weights):
    """Compute weighted portfolio return."""
    return sum(weights[ticker] * ret for ticker, ret in zip(weights.keys(), args))


port_ret = weighted_portfolio_return(
    0.15, 0.22, -0.05,
    AAPL=0.40, MSFT=0.35, TSLA=0.25
)

print(f"\nWeighted Portfolio Return: {port_ret:.2%}")


# ─────────────────────────────────────────
# 4.2 Lambdas, Higher-Order Functions & Closures
# ─────────────────────────────────────────

pct_change = lambda prev, curr: (curr - prev) / prev
log_return = lambda prev, curr: math.log(curr / prev)
annualize  = lambda r, n=252: (1 + r) ** n - 1

print(f"Simple return:   {pct_change(100, 112):.2%}")
print(f"Log return:      {log_return(100, 112):.4f}")
print(f"Annualized 0.1%: {annualize(0.001):.2%}")


stocks = [
    {"ticker": "AAPL", "market_cap": 2.8e12, "pe": 28.5},
    {"ticker": "MSFT", "market_cap": 3.1e12, "pe": 35.2},
    {"ticker": "GOOGL", "market_cap": 1.9e12, "pe": 25.1},
    {"ticker": "NVDA", "market_cap": 2.2e12, "pe": 65.8},
]

by_pe = sorted(stocks, key=lambda x: x["pe"])

print("\nStocks sorted by P/E ratio:")
for s in by_pe:
    print(f"  {s['ticker']}: P/E = {s['pe']}")


# ─────────────────────────────────────────
# Closures — SMA factory
# ─────────────────────────────────────────

def make_sma(window: int):
    history = []

    def update(price: float):
        history.append(price)
        if len(history) >= window:
            return sum(history[-window:]) / window
        return None

    return update


sma5 = make_sma(5)
sma10 = make_sma(10)

prices_stream = [100, 102, 101, 105, 103, 107, 110, 108, 112, 115]

print("\nStreaming SMA computation:")
for p in prices_stream:
    s5 = sma5(p)
    s10 = sma10(p)

    s5_str = f"{s5:.2f}" if s5 is not None else "--"
    s10_str = f"{s10:.2f}" if s10 is not None else "--"

    print(f"  Price={p:>5} | SMA5={s5_str} | SMA10={s10_str}")
```

## Module 5: OOP — Core Concepts

### Classes, Attributes, Methods, Encapsulation

💡 **OOP in Quant Finance:** Every financial instrument, strategy, or risk model can be modelled as an object with state and behaviour. OOP makes your code reusable, testable, and maintainable.

```python
# ─────────────────────────────────────────
# 5.1 Basic Class — Financial Instrument
# ─────────────────────────────────────────

class Stock:
    """
    Represents a publicly traded equity.
    Demonstrates: attributes, methods, encapsulation, class/static methods.
    """

    asset_class = "Equity"
    _instance_count = 0

    def __init__(self, ticker: str, name: str, price: float, shares_outstanding: float):
        self.ticker = ticker.upper()
        self.name = name
        self._price = price
        self.__shares = shares_outstanding
        self._price_history = [price]

        Stock._instance_count += 1

    def market_cap(self) -> float:
        return self._price * self.__shares

    def update_price(self, new_price: float) -> None:
        if new_price <= 0:
            raise ValueError(f"Price must be positive, got {new_price}")
        self._price_history.append(new_price)
        self._price = new_price

    def total_return(self) -> float:
        return (self._price - self._price_history[0]) / self._price_history[0]

    def price_history(self) -> list:
        return self._price_history.copy()

    @classmethod
    def get_instance_count(cls) -> int:
        return cls._instance_count

    @staticmethod
    def is_valid_ticker(ticker: str) -> bool:
        return ticker.isalpha() and 1 <= len(ticker) <= 5

    def __repr__(self) -> str:
        return f"Stock(ticker='{self.ticker}', price={self._price:.2f})"

    def __str__(self) -> str:
        return f"{self.name} ({self.ticker}): ${self._price:.2f}"


aapl = Stock("AAPL", "Apple Inc.", 182.45, 15_550_000_000)
msft = Stock("MSFT", "Microsoft Corp.", 415.30, 7_430_000_000)
googl = Stock("GOOGL", "Alphabet Inc.", 172.80, 12_800_000_000)

print(repr(aapl))
print(str(msft))
print(f"Market Cap AAPL: ${aapl.market_cap()/1e12:.2f}T")
print(f"Market Cap MSFT: ${msft.market_cap()/1e12:.2f}T")

aapl.update_price(190.00)
aapl.update_price(195.50)

print(f"\nAAPL total return: {aapl.total_return():.2%}")
print(f"AAPL price history: {aapl.price_history()}")

print(f"\nStocks created: {Stock.get_instance_count()}")
print(f"Valid ticker 'AAPL': {Stock.is_valid_ticker('AAPL')}")
print(f"Valid ticker 'TOOLONG': {Stock.is_valid_ticker('TOOLONG')}")
print(f"Asset class: {Stock.asset_class}")
```

## Module 6: OOP — Inheritance

### Single Inheritance, Multiple Inheritance, MRO

```python
# ─────────────────────────────────────────
# 6.1 Inheritance Hierarchy — Financial Instruments
# ─────────────────────────────────────────

class FinancialInstrument:
    """Base class for all financial instruments."""

    def __init__(self, name: str, currency: str = "USD"):
        self.name = name
        self.currency = currency

    def fair_value(self) -> float:
        raise NotImplementedError("Subclasses must implement fair_value()")

    def describe(self) -> str:
        return f"{self.__class__.__name__}: {self.name} ({self.currency})"

    def __repr__(self):
        return f"{self.__class__.__name__}('{self.name}')"


class Equity(FinancialInstrument):

    def __init__(self, name: str, ticker: str, price: float, currency: str = "USD"):
        super().__init__(name, currency)
        self.ticker = ticker
        self.price = price

    def fair_value(self) -> float:
        return self.price

    def dividend_yield(self, annual_dividend: float) -> float:
        return annual_dividend / self.price

    def describe(self) -> str:
        base = super().describe()
        return f"{base} | Ticker: {self.ticker} | Price: ${self.price:.2f}"


class Bond(FinancialInstrument):

    def __init__(self, name: str, face_value: float, coupon: float,
                 maturity: int, ytm: float, currency: str = "USD"):
        super().__init__(name, currency)
        self.face_value = face_value
        self.coupon = coupon
        self.maturity = maturity
        self.ytm = ytm

    def fair_value(self) -> float:
        c = self.face_value * self.coupon
        fv = self.face_value
        n = self.maturity
        y = self.ytm

        return c * (1 - (1 + y) ** -n) / y + fv / (1 + y) ** n



class ConvertibleBond(Bond):

    def __init__(self, name: str, face_value: float, coupon: float,
                 maturity: int, ytm: float, ticker: str,
                 stock_price: float, conversion_ratio: float):

        super().__init__(name, face_value, coupon, maturity, ytm)

        self.ticker = ticker
        self.stock_price = stock_price
        self.conversion_ratio = conversion_ratio

    @property
    def conversion_value(self):
        return self.conversion_ratio * self.stock_price
      
    def fair_value(self) -> float:
        return max(Bond.fair_value(self), self.conversion_value)

    def should_convert(self) -> bool:
        return self.conversion_value > Bond.fair_value(self)
  

# ─────────────────────────────────────────
# Usage
# ─────────────────────────────────────────

aapl_stock = Equity("Apple Inc.", "AAPL", 182.45)
us_treasury = Bond("US 10Y Treasury", 1000, 0.045, 10, 0.04)

conv_bond = ConvertibleBond(
    "TSLA 2% 2026",
    face_value=1000,
    coupon=0.02,
    maturity=2,
    ytm=0.03,
    ticker="TSLA",
    stock_price=185.50,
    conversion_ratio=5
)

instruments = [aapl_stock, us_treasury, conv_bond]

print("Financial Instruments:")
for inst in instruments:
    print(f"  {repr(inst):<30} | Fair Value: ${inst.fair_value():>10,.2f}")

print(f"\nConvertible bond should convert: {conv_bond.should_convert()}")
print(f"Conversion value: ${conv_bond.conversion_value:.2f}")
print(f"Bond value: ${Bond.fair_value(conv_bond):.2f}")

print("\nMRO for ConvertibleBond:")
for cls in ConvertibleBond.__mro__:
    print(f"  → {cls.__name__}")
```

## Module 7: OOP — Polymorphism

### Abstract Classes, Duck Typing, Method Overriding

```python
from abc import ABC, abstractmethod
import math, statistics
import random


class RiskModel(ABC):
    """
    Abstract base class for risk models.
    All concrete risk models MUST implement these methods.
    """
    
    def __init__(self, returns: list, confidence: float = 0.95):
        if not returns:
            raise ValueError("Returns list cannot be empty")
        self.returns    = returns
        self.confidence = confidence
    
    @abstractmethod
    def var(self) -> float:
        """Compute Value-at-Risk."""
        pass
    
    @abstractmethod
    def cvar(self) -> float:
        """Compute Conditional VaR (Expected Shortfall)."""
        pass
    
    @abstractmethod
    def model_name(self) -> str:
        pass
    
    # Concrete method (shared by all subclasses)
    def volatility(self) -> float:
        return statistics.stdev(self.returns)
    
    def report(self) -> str:
        return (
            f"\n{'─'*45}\n"
            f"  Risk Model : {self.model_name()}\n"
            f"  Confidence : {self.confidence:.0%}\n"
            f"  Volatility : {self.volatility():.4f}\n"
            f"  VaR        : {self.var():.4f} ({self.var()*100:.2f}%)\n"
            f"  CVaR       : {self.cvar():.4f} ({self.cvar()*100:.2f}%)\n"
            f"{'─'*45}"
        )


class HistoricalVaR(RiskModel):
    """Historical Simulation VaR."""
    
    def model_name(self) -> str:
        return "Historical Simulation"
    
    def var(self) -> float:
        sorted_rets = sorted(self.returns)
        idx = int((1 - self.confidence) * len(sorted_rets))
        return abs(sorted_rets[idx])
    
    def cvar(self) -> float:
        sorted_rets = sorted(self.returns)
        idx = int((1 - self.confidence) * len(sorted_rets))
        tail = sorted_rets[:idx+1]
        return abs(sum(tail) / len(tail))


class ParametricVaR(RiskModel):
    """Parametric (Normal Distribution) VaR."""
    
    # z-score lookup (simplified)
    _Z_SCORES = {0.90: 1.282, 0.95: 1.645, 0.99: 2.326}
    
    def model_name(self) -> str:
        return "Parametric (Gaussian)"
    
    def var(self) -> float:
        z   = self._Z_SCORES.get(self.confidence, 1.645)
        mu  = statistics.mean(self.returns)
        std = self.volatility()
        return abs(mu - z * std)
    
    def cvar(self) -> float:
        """CVaR for normal distribution: E[L | L > VaR]."""
        z   = self._Z_SCORES.get(self.confidence, 1.645)
        std = self.volatility()
        mu  = statistics.mean(self.returns)
        phi = math.exp(-z**2 / 2) / math.sqrt(2 * math.pi)
        return abs(mu - std * phi / (1 - self.confidence))


# ── Polymorphism in action ──
random.seed(42)
portfolio_returns = [random.gauss(-0.0002, 0.015) for _ in range(500)]

models = [
    HistoricalVaR(portfolio_returns, confidence=0.95),
    HistoricalVaR(portfolio_returns, confidence=0.99),
    ParametricVaR(portfolio_returns, confidence=0.95),
    ParametricVaR(portfolio_returns, confidence=0.99),
]

# Same interface, different behaviour — polymorphism!
for model in models:
    print(model.report())

# Can't instantiate abstract class
try:
    r = RiskModel(portfolio_returns)
except TypeError as e:
    print(f"\nCannot instantiate ABC: {e}")
```

Dunder Methods, Properties, Descriptors

```python
class Portfolio:
    """
    A portfolio of holdings with rich operator support.
    Demonstrates comprehensive dunder (magic) methods.
    """
    
    def __init__(self, name: str, holdings: dict = None):
        """
        Args:
            holdings: {ticker: {"shares": int, "avg_price": float}} mapping
        """
        self.name     = name
        self.holdings = holdings or {}
    
    # ── Container protocol ──
    def __len__(self) -> int:
        """Number of positions."""
        return len(self.holdings)
    
    def __contains__(self, ticker: str) -> bool:
        """'AAPL' in portfolio → True/False."""
        return ticker in self.holdings
    
    def __getitem__(self, ticker: str):
        """portfolio['AAPL'] → position dict."""
        return self.holdings[ticker]
    
    def __setitem__(self, ticker: str, position: tuple):
        """portfolio['NVDA'] = (100, 450.0)."""
        shares, price = position
        self.holdings[ticker] = {"shares": shares, "avg_price": price}
    
    def __delitem__(self, ticker: str):
        """del portfolio['TSLA']"""
        del self.holdings[ticker]
    
    def __iter__(self):
        """Iterate over tickers."""
        return iter(self.holdings)
    
    # ── Arithmetic operators ──
    def __add__(self, other: "Portfolio") -> "Portfolio":
        """Merge two portfolios."""
        merged = {**self.holdings}
        for ticker, pos in other.holdings.items():
            if ticker in merged:
                # Weighted average cost
                s1, p1 = merged[ticker]["shares"], merged[ticker]["avg_price"]
                s2, p2 = pos["shares"], pos["avg_price"]
                total_shares = s1 + s2
                avg_price    = (s1*p1 + s2*p2) / total_shares
                merged[ticker] = {"shares": total_shares, "avg_price": avg_price}
            else:
                merged[ticker] = pos
        return Portfolio(f"{self.name}+{other.name}", merged)
    
    def __mul__(self, scalar: float) -> "Portfolio":
        """Scale all positions."""
        scaled = {t: {"shares": p["shares"]*scalar, "avg_price": p["avg_price"]}
                  for t, p in self.holdings.items()}
        return Portfolio(f"{self.name}×{scalar}", scaled)
    
    # ── Comparison operators ──
    def total_cost(self) -> float:
        return sum(p["shares"] * p["avg_price"] for p in self.holdings.values())
    
    def __lt__(self, other): 
        return self.total_cost() < other.total_cost()
    
    def __le__(self, other): 
        return self.total_cost() <= other.total_cost()
    
    def __eq__(self, other): 
        return self.total_cost() == other.total_cost()
    
    def __gt__(self, other): 
        return self.total_cost() > other.total_cost()
    
    # ── Representation ──
    def __repr__(self) -> str:
        return f"Portfolio('{self.name}', {len(self)} positions, cost=${self.total_cost():,.0f})"
    
    def __str__(self) -> str:
        lines = [f"Portfolio: {self.name}"]
        lines.append(f"{'Ticker':<8} {'Shares':>8} {'Avg Price':>12} {'Cost Basis':>14}")
        lines.append("-" * 45)
        for ticker, pos in self.holdings.items():
            cost = pos['shares'] * pos['avg_price']
            lines.append(f"{ticker:<8} {pos['shares']:>8,.0f} ${pos['avg_price']:>11.2f} ${cost:>13,.2f}")
        lines.append("-" * 45)
        lines.append(f"{'TOTAL':<8} {'':>8} {'':>12} ${self.total_cost():>13,.2f}")
        return "\n".join(lines)


# ─────────────────────────────────────────
# Example Usage
# ─────────────────────────────────────────

# Build portfolios
p1 = Portfolio("Tech", {
    "AAPL": {"shares": 100, "avg_price": 150.0},
    "MSFT": {"shares": 50,  "avg_price": 380.0},
})

p2 = Portfolio("Growth", {
    "NVDA": {"shares": 30,  "avg_price": 600.0},
    "AAPL": {"shares": 50,  "avg_price": 175.0},  # AAPL in both!
})

print(str(p1))

# Dunder demonstrations
print(f"\nPositions in p1: {len(p1)}")
print(f"'AAPL' in p1: {'AAPL' in p1}")
print(f"'TSLA' in p1: {'TSLA' in p1}")

# Add a new position
p1["GOOGL"] = (30, 140.0)
print(f"After adding GOOGL: {len(p1)} positions")

# Merge portfolios
combined = p1 + p2
print(f"\nCombined: {repr(combined)}")

# Scale portfolio
half = p1 * 0.5
print(f"Half of p1: {repr(half)}")

# Comparisons
print(f"\np1 > p2: {p1 > p2}")
print(f"p1 < p2: {p1 < p2}")

# Iteration
print("\nTickers in p1:")
for ticker in p1:
    print(f"  {ticker}")
```

### Singleton, Factory, Observer, Strategy

```python
# ─────────────────────────────────────────
# 9.1 Singleton — Market Data Feed
# ─────────────────────────────────────────

class MarketDataFeed:
    _instance = None
    _prices = {}

    def __new__(cls):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
            print("📡 Market data feed connected (new instance)")
        return cls._instance

    def update(self, ticker: str, price: float):
        self._prices[ticker] = price

    def get(self, ticker: str) -> float:
        return self._prices.get(ticker, 0.0)


feed1 = MarketDataFeed()
feed2 = MarketDataFeed()

feed1.update("AAPL", 182.45)

print(f"Same instance: {feed1 is feed2}")
print(f"AAPL via feed2: ${feed2.get('AAPL')}")


# ─────────────────────────────────────────
# 9.2 Factory — Instrument Creator
# ─────────────────────────────────────────

class InstrumentFactory:

    @staticmethod
    def create(config: dict):
        kind = config.get("type", "").lower()

        if kind == "equity":
            return Equity(config["name"], config["ticker"], config["price"])

        elif kind == "bond":
            return Bond(
                config["name"],
                config["face"],
                config["coupon"],
                config["maturity"],
                config["ytm"]
            )

        else:
            raise ValueError(f"Unknown instrument type: {kind}")


configs = [
    {"type": "equity", "name": "Apple", "ticker": "AAPL", "price": 182.45},
    {"type": "bond", "name": "US T-Note", "face": 1000, "coupon": 0.04, "maturity": 5, "ytm": 0.035},
]

print("\nFactory-created instruments:")
for cfg in configs:
    inst = InstrumentFactory.create(cfg)
    print(f"  {repr(inst)} → Fair value: ${inst.fair_value():.2f}")


# ─────────────────────────────────────────
# 9.3 Observer — Price Alert System
# ─────────────────────────────────────────

from abc import ABC, abstractmethod


class PriceObserver(ABC):

    @abstractmethod
    def on_price_update(self, ticker: str, price: float):
        pass


class StopLossAlert(PriceObserver):

    def __init__(self, ticker: str, stop: float):
        self.ticker = ticker
        self.stop = stop

    def on_price_update(self, ticker: str, price: float):
        if ticker == self.ticker and price < self.stop:
            print(f"  🚨 STOP-LOSS: {ticker} @ ${price:.2f} < ${self.stop:.2f}")


class PriceLogger(PriceObserver):

    def on_price_update(self, ticker: str, price: float):
        print(f"  📝 LOG: {ticker} updated to ${price:.2f}")


class PriceFeedSubject:

    def __init__(self):
        self._observers = []

    def subscribe(self, obs: PriceObserver):
        self._observers.append(obs)

    def unsubscribe(self, obs: PriceObserver):
        self._observers.remove(obs)

    def notify(self, ticker: str, price: float):
        for obs in self._observers:
            obs.on_price_update(ticker, price)


feed = PriceFeedSubject()
feed.subscribe(StopLossAlert("AAPL", 175.0))
feed.subscribe(PriceLogger())

print("\nObserver pattern — price stream:")
for price in [182.0, 178.5, 173.0, 171.5]:
    feed.notify("AAPL", price)


# ─────────────────────────────────────────
# 9.4 Strategy — Execution Algorithm
# ─────────────────────────────────────────

class ExecutionStrategy(ABC):

    @abstractmethod
    def execute(self, ticker: str, qty: int, prices: list):
        pass


class TWAP(ExecutionStrategy):

    def execute(self, ticker, qty, prices):
        slice_qty = qty // len(prices)
        return [(slice_qty, p) for p in prices]


class VWAP(ExecutionStrategy):

    def execute(self, ticker, qty, prices):
        min_p = min(prices)
        weights = [min_p / p for p in prices]
        total_w = sum(weights)
        return [(int(qty * w / total_w), p) for w, p in zip(weights, prices)]


class OrderExecutor:

    def __init__(self, strategy: ExecutionStrategy):
        self.strategy = strategy

    def set_strategy(self, strategy: ExecutionStrategy):
        self.strategy = strategy

    def run(self, ticker: str, qty: int, prices: list):
        fills = self.strategy.execute(ticker, qty, prices)
        total_qty = sum(q for q, _ in fills)
        total_cost = sum(q * p for q, p in fills)
        avg_price = total_cost / total_qty if total_qty else 0

        name = self.strategy.__class__.__name__
        print(f"  [{name}] Filled {total_qty} @ avg ${avg_price:.2f}")


intraday_prices = [182.0, 181.5, 183.0, 184.5, 182.8]

executor = OrderExecutor(TWAP())
print("\nExecution strategies:")
executor.run("AAPL", 1000, intraday_prices)

executor.set_strategy(VWAP())
executor.run("AAPL", 1000, intraday_prices)
```

## Module 10: Error Handling

### try/except/finally, Custom Exceptions

```python
# ─────────────────────────────────────────
# 10.1 Custom Exception Hierarchy
# ─────────────────────────────────────────

class QuantFinanceError(Exception):
    """Base exception for all quant finance errors."""
    pass


class InsufficientFundsError(QuantFinanceError):

    def __init__(self, required: float, available: float):
        self.required = required
        self.available = available
        super().__init__(f"Need ${required:,.2f}, only ${available:,.2f} available")


class InvalidOrderError(QuantFinanceError):

    def __init__(self, reason: str, order_details: dict):
        self.order_details = order_details
        super().__init__(f"Invalid order: {reason} | Details: {order_details}")


class MarketClosedError(QuantFinanceError):
    pass


class RiskLimitBreachError(QuantFinanceError):

    def __init__(self, metric: str, value: float, limit: float):
        super().__init__(f"Risk limit breached: {metric}={value:.4f} > limit={limit:.4f}")


# ─────────────────────────────────────────
# 10.2 Exception Handling — Order System
# ─────────────────────────────────────────

class OrderManagementSystem:

    def __init__(self, cash: float, max_position_pct: float = 0.20):
        self.cash = cash
        self.max_position_pct = max_position_pct
        self.portfolio_value = cash
        self._trade_log = []

    def submit_order(self, ticker: str, qty: int, price: float, side: str) -> str:
        order = {"ticker": ticker, "qty": qty, "price": price, "side": side}

        try:
            if qty <= 0:
                raise InvalidOrderError("Quantity must be positive", order)

            if price <= 0:
                raise InvalidOrderError("Price must be positive", order)

            if side not in ("BUY", "SELL"):
                raise InvalidOrderError(f"Unknown side '{side}'", order)

            order_value = qty * price

            if side == "BUY":
                if order_value > self.cash:
                    raise InsufficientFundsError(order_value, self.cash)

                pct = order_value / self.portfolio_value
                if pct > self.max_position_pct:
                    raise RiskLimitBreachError("position_pct", pct, self.max_position_pct)

                self.cash -= order_value

            else:
                self.cash += order_value

            status = f"FILLED: {side} {qty} {ticker} @ ${price:.2f}"
            self._trade_log.append(status)
            return status

        except InsufficientFundsError as e:
            return f"REJECTED (funds): {e}"

        except RiskLimitBreachError as e:
            return f"REJECTED (risk): {e}"

        except InvalidOrderError as e:
            return f"REJECTED (invalid): {e}"

        except QuantFinanceError as e:
            return f"REJECTED (quant error): {e}"

        except Exception as e:
            print(f"UNEXPECTED ERROR: {type(e).__name__}: {e}")
            raise

        finally:
            pass  # audit logging hook


oms = OrderManagementSystem(cash=50_000, max_position_pct=0.20)

print("Order Management System:")
print(f"  Starting cash: ${oms.cash:,.2f}")

test_orders = [
    ("AAPL", 50, 182.45, "BUY"),
    ("MSFT", 200, 415.30, "BUY"),
    ("GOOGL", 100, 172.80, "BUY"),
    ("TSLA", -10, 185.0, "BUY"),
    ("NVDA", 10, 620.0, "SELL"),
]

for ticker, qty, price, side in test_orders:
    result = oms.submit_order(ticker, qty, price, side)
    print(f"  {result}")

print(f"\n  Remaining cash: ${oms.cash:,.2f}")
```

#### Module 14: Decorators

```python
import time, functools
import math
import random
from collections import namedtuple

# ── Function decorator: execution timer ──
def timer(func):
    """Decorator: measure execution time of a function."""
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        start  = time.perf_counter()
        result = func(*args, **kwargs)
        end    = time.perf_counter()
        print(f"  ⏱  {func.__name__}() executed in {(end-start)*1000:.3f} ms")
        return result
    return wrapper

# ── Decorator with arguments: retry on failure ──
def retry(max_attempts: int = 3, delay: float = 0.01):
    """Retry decorator — useful for flaky data feeds."""
    def decorator(func):
        @functools.wraps(func)
        def wrapper(*args, **kwargs):
            for attempt in range(1, max_attempts + 1):
                try:
                    return func(*args, **kwargs)
                except Exception as e:
                    print(f"  Attempt {attempt}/{max_attempts} failed: {e}")
                    if attempt < max_attempts:
                        time.sleep(delay)
            raise RuntimeError(f"{func.__name__} failed after {max_attempts} attempts")
        return wrapper
    return decorator

# ── Memoization: cache expensive computations ──
def memoize(func):
    cache = {}
    @functools.wraps(func)
    def wrapper(*args):
        if args not in cache:
            cache[args] = func(*args)
        return cache[args]
    return wrapper

# ── Class decorator: add logging to all methods ──
def log_all_methods(cls):
    for attr_name in dir(cls):
        if not attr_name.startswith("_"):
            method = getattr(cls, attr_name)
            if callable(method):
                setattr(cls, attr_name, timer(method))
    return cls


@timer
def compute_covariance_matrix(returns_matrix: list) -> list:
    """Compute covariance matrix from returns (O(n²) for demonstration)."""
    n_assets = len(returns_matrix)
    n_obs    = len(returns_matrix[0])
    means    = [sum(r)/n_obs for r in returns_matrix]
    cov = [[0.0]*n_assets for _ in range(n_assets)]
    for i in range(n_assets):
        for j in range(i, n_assets):
            c = sum((returns_matrix[i][k]-means[i]) * (returns_matrix[j][k]-means[j])
                    for k in range(n_obs)) / (n_obs - 1)
            cov[i][j] = cov[j][i] = c
    return cov


@memoize
def black_scholes_d1(S, K, T, r, sigma):
    """Memoized d1 component of Black-Scholes formula."""
    return (math.log(S/K) + (r + 0.5*sigma**2)*T) / (sigma * math.sqrt(T))


# Simulate returns
random.seed(0)
ret_matrix = [[random.gauss(0.001, 0.02) for _ in range(100)] for _ in range(4)]

print("Decorator demonstrations:")
cov = compute_covariance_matrix(ret_matrix)
print(f"  Covariance [0,0]: {cov[0][0]:.6f}")

# Memoize: second call is instant
d1a = black_scholes_d1(150, 155, 0.5, 0.05, 0.20)
d1b = black_scholes_d1(150, 155, 0.5, 0.05, 0.20)  # from cache
print(f"\n  BS d1 = {d1a:.4f} (both calls return same: {d1a == d1b})")


# functools.lru_cache (built-in memoization)
@functools.lru_cache(maxsize=256)
def binomial_coeff(n: int, k: int) -> int:
    """Binomial coefficient — used in binomial option pricing."""
    if k == 0 or k == n: return 1
    return binomial_coeff(n-1, k-1) + binomial_coeff(n-1, k)

print(f"\n  C(10,5) = {binomial_coeff(10,5)} (binomial option tree node)")


@retry(max_attempts=3)
def fetch_price(ticker: str) -> float:
    """Simulated flaky data feed."""
    if random.random() < 0.5:
        raise ConnectionError("Feed timeout")
    return 182.45

try:
    price = fetch_price("AAPL")
    print(f"\n  Fetched AAPL: ${price}")
except RuntimeError as e:
    print(f"\n  {e}")
```

#### Module 15: Context Managers

```python
from contextlib import contextmanager
import time

class InsufficientFundsError(Exception):
    pass

class RiskLimitBreachError(Exception):
    pass

class TradingSession:
    """
    Context manager for a trading session.
    Ensures clean entry/exit: connect, trade, disconnect.
    """
    
    def __init__(self, session_id: str, cash: float):
        self.session_id = session_id
        self.cash       = cash
        self.trades     = []
        self._start_time = None
    
    def __enter__(self):
        self._start_time = time.time()
        print(f"\n🔌 Session '{self.session_id}' opened | Cash: ${self.cash:,.2f}")
        return self   # `as` target
    
    def __exit__(self, exc_type, exc_val, exc_tb):
        elapsed = time.time() - self._start_time
        if exc_type:
            print(f"⚠️  Session ended with error: {exc_type.__name__}: {exc_val}")
        else:
            print(f"✅ Session '{self.session_id}' closed | Trades: {len(self.trades)} | Elapsed: {elapsed:.3f}s")
        return False   # do not suppress exceptions
    
    def buy(self, ticker: str, qty: int, price: float):
        cost = qty * price
        if cost > self.cash:
            raise InsufficientFundsError(cost, self.cash)
        self.cash -= cost
        self.trades.append((ticker, "BUY", qty, price))
        print(f"  BUY  {qty:>5} {ticker} @ ${price:.2f} | Remaining cash: ${self.cash:,.2f}")


# ── Class-based context manager ──
with TradingSession("ALGO_001", cash=50_000) as session:
    session.buy("AAPL", 50, 182.45)
    session.buy("MSFT", 20, 415.30)


# ── Generator-based context manager ──
@contextmanager
def risk_monitor(portfolio_value: float, max_drawdown_pct: float = 0.10):
    """Monitor portfolio drawdown; raise if limit breached."""
    print(f"\n📊 Risk monitor active | Max DD: {max_drawdown_pct:.0%}")
    try:
        yield
    except RiskLimitBreachError:
        print("  🚨 Risk monitor intercepted breach — flattening positions")
        raise
    finally:
        print("  📊 Risk monitor deactivated")


with risk_monitor(100_000, 0.10):
    print("  Trading within risk limits...")
    print("  All trades completed safely.")
```

```python
from collections import namedtuple
import math
from functools import reduce, partial
from itertools import accumulate

# ── Custom Iterator ──
class PriceBarIterator:
    """
    Custom iterator for OHLCV bars.
    Implements the iterator protocol: __iter__ and __next__.
    """
    
    def __init__(self, bars: list):
        self._bars  = bars
        self._index = 0
    
    def __iter__(self):
        return self
    
    def __next__(self):
        if self._index >= len(self._bars):
            raise StopIteration
        bar = self._bars[self._index]
        self._index += 1
        return bar
    
    def __len__(self):
        return len(self._bars)


OHLCV = namedtuple("OHLCV", ["date", "open", "high", "low", "close", "volume"])
sample_bars = [
    OHLCV("2024-01-02", 185.02, 185.09, 182.13, 184.21, 55_759_600),
    OHLCV("2024-01-03", 183.92, 185.15, 182.73, 184.25, 49_895_800),
    OHLCV("2024-01-04", 182.15, 183.09, 180.63, 181.91, 47_673_900),
]

bar_iter = PriceBarIterator(sample_bars)
print("Custom bar iterator:")
for bar in bar_iter:
    rng = bar.high - bar.low
    print(f"  {bar.date} | Close: ${bar.close} | Range: ${rng:.2f} | Vol: {bar.volume:,}")


# ── Functional Programming ──
ret_list = [0.02, -0.01, 0.03, 0.015, -0.005, 0.025, 0.01]

# map
log_rets = list(map(lambda r: math.log(1 + r), ret_list))
print(f"\nLog returns: {[round(r,4) for r in log_rets]}")

# filter
positive = list(filter(lambda r: r > 0, ret_list))
print(f"Positive only: {[round(r,4) for r in positive]}")

# reduce
compound = reduce(lambda acc, r: acc * (1 + r), ret_list, 1.0) - 1
print(f"Compound return: {compound:.2%}")

# accumulate
cum_returns = list(accumulate(ret_list, lambda acc, r: acc * (1 + r), initial=1.0))
print(f"Cumulative values: {[round(v,4) for v in cum_returns]}")

# partial
def kelly_fraction(win_prob: float, win_return: float, loss_return: float) -> float:
    """Kelly Criterion: optimal bet fraction."""
    return win_prob / loss_return - (1 - win_prob) / win_return

strategy_kelly = partial(kelly_fraction, 0.55)
print(f"\nKelly fraction: {strategy_kelly(0.10, 0.08):.2%}")
```

### Module 18: Type Hints & Dataclasses

```python
from dataclasses import dataclass, field, asdict
from datetime import datetime
import math
from typing import Optional, ClassVar

# ─────────────────────────────────────────
# 18. Type Hints, Dataclasses, TypedDict
# ─────────────────────────────────────────

@dataclass
class MarketOrder:
    """Immutable market order."""
    ticker     : str
    quantity   : int
    side       : str                          # 'BUY' or 'SELL'
    price      : float
    timestamp  : str = field(default_factory=lambda: datetime.now().isoformat())
    order_id   : Optional[str] = None
    
    # Validation via __post_init__
    def __post_init__(self):
        if self.quantity <= 0:
            raise ValueError("Quantity must be positive")
        if self.side not in ("BUY", "SELL"):
            raise ValueError(f"Invalid side: {self.side}")
        self.side = self.side.upper()
    
    @property
    def notional(self) -> float:
        return self.quantity * self.price


@dataclass(order=True, frozen=True)
class PriceLevel:
    """Order book price level — frozen (immutable) and orderable."""
    price     : float
    quantity  : int = field(compare=False)
    side      : str = field(compare=False, default="BID")


@dataclass
class OptionContract:
    """Vanilla option with Black-Scholes pricing."""
    underlying   : str
    spot         : float
    strike       : float
    maturity     : float   # in years
    risk_free    : float
    volatility   : float
    option_type  : str = "call"  # 'call' or 'put'
    
    # Class variable (not per-instance)
    SQRT_2PI: ClassVar[float] = math.sqrt(2 * math.pi)
    
    def _norm_cdf(self, x: float) -> float:
        """Approximate standard normal CDF."""
        return 0.5 * (1 + math.erf(x / math.sqrt(2)))
    
    def _d1_d2(self):
        S, K, T = self.spot, self.strike, self.maturity
        r, σ    = self.risk_free, self.volatility
        d1 = (math.log(S/K) + (r + 0.5*σ**2)*T) / (σ * math.sqrt(T))
        d2 = d1 - σ * math.sqrt(T)
        return d1, d2
    
    def price(self) -> float:
        """Black-Scholes price."""
        S, K = self.spot, self.strike
        r, T = self.risk_free, self.maturity
        d1, d2 = self._d1_d2()
        N = self._norm_cdf
        if self.option_type == "call":
            return S * N(d1) - K * math.exp(-r*T) * N(d2)
        else:
            return K * math.exp(-r*T) * N(-d2) - S * N(-d1)
    
    def delta(self) -> float:
        """Option delta."""
        d1, _ = self._d1_d2()
        if self.option_type == "call":
            return self._norm_cdf(d1)
        return self._norm_cdf(d1) - 1
    
    def gamma(self) -> float:
        """Option gamma."""
        d1, _ = self._d1_d2()
        S, σ, T = self.spot, self.volatility, self.maturity
        phi = math.exp(-d1**2/2) / self.SQRT_2PI
        return phi / (S * σ * math.sqrt(T))
    
    def vega(self) -> float:
        """Option vega (per 1% change in vol)."""
        d1, _ = self._d1_d2()
        S, T = self.spot, self.maturity
        phi = math.exp(-d1**2/2) / self.SQRT_2PI
        return S * phi * math.sqrt(T) * 0.01  # per 1%


# ── Usage ──
order = MarketOrder(ticker="AAPL", quantity=100, side="BUY", price=182.45)
print(f"Order: {order}")
print(f"Notional: ${order.notional:,.2f}")

# Order book
bids = [PriceLevel(182.40, 500, "BID"), 
        PriceLevel(182.45, 200, "BID"), 
        PriceLevel(182.35, 1000, "BID")]
best_bid = max(bids)  # works because dataclass(order=True)
print(f"\nBest bid: ${best_bid.price} x {best_bid.quantity}")

# Black-Scholes option pricing
call = OptionContract("AAPL", spot=182.45, strike=185.0, maturity=0.25, 
                      risk_free=0.05, volatility=0.25, option_type="call")
put  = OptionContract("AAPL", spot=182.45, strike=185.0, maturity=0.25,
                      risk_free=0.05, volatility=0.25, option_type="put")

print(f"\n{'Black-Scholes Option Pricing':=^45}")
for opt in [call, put]:
    print(f"  {opt.option_type.upper():<5}: Price=${opt.price():.4f} | "
          f"Delta={opt.delta():.4f} | Gamma={opt.gamma():.4f} | Vega={opt.vega():.4f}")

# Put-call parity check
lhs = call.price() - put.price()
rhs = call.spot - call.strike * math.exp(-call.risk_free * call.maturity)
print(f"\nPut-Call Parity: C-P={lhs:.4f} ≈ S-Ke^(-rT)={rhs:.4f} ✓")
print(f"\nOrder as dict: {asdict(order)}")
```

```python
import threading
import multiprocessing
import asyncio
import time
import random
from concurrent.futures import ThreadPoolExecutor

# ─────────────────────────────────────────
# 21. Concurrency — Parallel Monte Carlo & Market Data
# ─────────────────────────────────────────

# ── Threading: Thread-safe portfolio updater ──
class ThreadSafePortfolioUpdater:
    """Thread-safe portfolio updater using locks."""
    
    def __init__(self):
        self._prices = {}
        self._lock   = threading.Lock()
    
    def update(self, ticker: str, price: float):
        with self._lock:   # mutual exclusion
            self._prices[ticker] = price
    
    def get_snapshot(self) -> dict:
        with self._lock:
            return self._prices.copy()


def simulate_price_feed(updater, ticker, base_price, n_updates=5):
    """Simulates a streaming price feed for one ticker."""
    price = base_price
    for _ in range(n_updates):
        price *= (1 + random.gauss(0, 0.001))
        updater.update(ticker, round(price, 2))
        time.sleep(0.001)


# Run threaded price feeds
updater = ThreadSafePortfolioUpdater()
feeds = [
    ("AAPL", 182.45), ("MSFT", 415.30), ("GOOGL", 172.80), ("TSLA", 185.50)
]

threads = [threading.Thread(target=simulate_price_feed, args=(updater, t, p))
           for t, p in feeds]

for thr in threads: 
    thr.start()
for thr in threads: 
    thr.join()

snapshot = updater.get_snapshot()
print("Threaded price feed snapshot:")
for ticker, price in snapshot.items():
    print(f"  {ticker}: ${price}")


# ── ThreadPoolExecutor: Parallel backtesting ──
def backtest_strategy(params: tuple) -> dict:
    """Simplified backtest for given parameters."""
    sma_short, sma_long = params
    random.seed(sma_short * 100 + sma_long)
    # Simulate backtest result
    sharpe = random.gauss(0.8, 0.3)
    return {"sma_short": sma_short, "sma_long": sma_long, "sharpe": round(sharpe, 4)}


param_grid = [(s, l) for s in [5, 10, 20] for l in [50, 100, 200]]

with ThreadPoolExecutor(max_workers=4) as executor:
    results = list(executor.map(backtest_strategy, param_grid))

best = max(results, key=lambda x: x["sharpe"])
print(f"\nGrid search results ({len(results)} combos):")
for r in sorted(results, key=lambda x: -x['sharpe'])[:3]:
    print(f"  SMA({r['sma_short']}/{r['sma_long']}) → Sharpe: {r['sharpe']}")
print(f"  Best: SMA({best['sma_short']}/{best['sma_long']}) Sharpe={best['sharpe']}")


# ── Async/Await: Asynchronous market data fetching ──
async def fetch_price_async(ticker: str, delay: float) -> tuple:
    """Simulate async price fetch (non-blocking)."""
    await asyncio.sleep(delay)   # non-blocking wait
    price = {"AAPL": 182.45, "MSFT": 415.30, "NVDA": 620.0}.get(ticker, 0.0)
    return ticker, price


async def fetch_all_prices(tickers: list) -> dict:
    tasks = [fetch_price_async(t, i*0.01) for i, t in enumerate(tickers)]
    results = await asyncio.gather(*tasks)
    return dict(results)


# Run async (in Jupyter/IPython use await, otherwise use asyncio.run())
prices_async = await fetch_all_prices(["AAPL", "MSFT", "NVDA"])
print(f"\nAsync fetched prices: {prices_async}")
```

### Module 23: Capstone — Full OOP Quantitative Trading System

```python
# ═══════════════════════════════════════════════════════════════════
# CAPSTONE: Quantitative Portfolio Engine with Risk Management
# Full Production-Grade Quantitative Trading System
# ═══════════════════════════════════════════════════════════════════

from dataclasses import dataclass, field
from abc import ABC, abstractmethod
from typing import Optional
import numpy as np
import pandas as pd
import math
from datetime import datetime

# ─────────────────────────────────────────
# LAYER 1: Data Model
# ─────────────────────────────────────────

@dataclass(frozen=True)
class Asset:
    """Immutable asset identifier."""
    ticker   : str
    name     : str
    sector   : str
    currency : str = "USD"
    
    def __post_init__(self):
        object.__setattr__(self, 'ticker', self.ticker.upper())


@dataclass
class Position:
    """Live trading position."""
    asset         : Asset
    shares        : float
    avg_cost      : float
    current_price : float = 0.0
    
    @property
    def market_value(self) -> float:
        return self.shares * self.current_price
    
    @property
    def cost_basis(self) -> float:
        return self.shares * self.avg_cost
    
    @property
    def unrealized_pnl(self) -> float:
        return self.market_value - self.cost_basis
    
    @property
    def pnl_pct(self) -> float:
        return (self.current_price - self.avg_cost) / self.avg_cost if self.avg_cost != 0 else 0.0


# ─────────────────────────────────────────
# LAYER 2: Signal Framework (Strategy Pattern)
# ─────────────────────────────────────────

class Signal(ABC):
    """Abstract signal generator."""
    
    @abstractmethod
    def compute(self, prices: pd.Series) -> float:
        """Return a signal in [-1, +1]. +1=full long, -1=full short."""
        pass
    
    @property
    @abstractmethod
    def name(self) -> str: pass


class MomentumSignal(Signal):
    """Price momentum over a lookback window."""
    
    def __init__(self, lookback: int = 20):
        self.lookback = lookback
    
    @property
    def name(self) -> str:
        return f"Momentum({self.lookback})"
    
    def compute(self, prices: pd.Series) -> float:
        if len(prices) < self.lookback:
            return 0.0
        momentum = (prices.iloc[-1] / prices.iloc[-self.lookback]) - 1
        return float(np.clip(momentum * 10, -1, 1))


class MeanReversionSignal(Signal):
    """Z-score based mean reversion signal."""
    
    def __init__(self, window: int = 20):
        self.window = window
    
    @property
    def name(self) -> str:
        return f"MeanReversion({self.window})"
    
    def compute(self, prices: pd.Series) -> float:
        if len(prices) < self.window:
            return 0.0
        recent   = prices.iloc[-self.window:]
        z_score  = (prices.iloc[-1] - recent.mean()) / recent.std()
        return float(np.clip(-z_score / 2, -1, 1))


class CompositeSignal(Signal):
    """Weighted combination of multiple signals."""
    
    def __init__(self, signals_weights: list):
        """Args: [(Signal, weight), ...]"""
        self.signals_weights = signals_weights
        total_w = sum(w for _, w in signals_weights)
        self.signals_weights = [(s, w/total_w) for s, w in signals_weights]
    
    @property
    def name(self) -> str:
        parts = [f"{s.name}×{w:.0%}" for s, w in self.signals_weights]
        return f"Composite({', '.join(parts)})"
    
    def compute(self, prices: pd.Series) -> float:
        return sum(w * s.compute(prices) for s, w in self.signals_weights)


# ─────────────────────────────────────────
# LAYER 3: Risk Engine
# ─────────────────────────────────────────

class RiskEngine:
    """Portfolio risk management engine."""
    
    def __init__(self, max_position_pct: float = 0.20,
                 max_drawdown: float = 0.15,
                 var_confidence: float = 0.95):
        self.max_position_pct = max_position_pct
        self.max_drawdown     = max_drawdown
        self.var_confidence   = var_confidence
    
    def historical_var(self, returns: list) -> float:
        if not returns: return 0.0
        sorted_r = sorted(returns)
        idx = int((1 - self.var_confidence) * len(sorted_r))
        return abs(sorted_r[max(idx, 0)])
    
    def historical_cvar(self, returns: list) -> float:
        if not returns: return 0.0
        sorted_r = sorted(returns)
        idx = int((1 - self.var_confidence) * len(sorted_r))
        tail = sorted_r[:max(idx+1, 1)]
        return abs(sum(tail) / len(tail))
    
    def current_drawdown(self, equity_curve: list) -> float:
        if not equity_curve: return 0.0
        peak = max(equity_curve)
        return (peak - equity_curve[-1]) / peak
    
    def check_position_limit(self, order_value: float, portfolio_value: float) -> bool:
        return (order_value / portfolio_value) <= self.max_position_pct
    
    def check_drawdown_limit(self, equity_curve: list) -> bool:
        return self.current_drawdown(equity_curve) < self.max_drawdown


# ─────────────────────────────────────────
# LAYER 4: Portfolio Engine (Core)
# ─────────────────────────────────────────

class PortfolioEngine:
    """Full quantitative portfolio engine."""
    
    VERSION = "1.0.0"
    
    def __init__(self, name: str, initial_capital: float,
                 signal: Signal, risk_engine: RiskEngine):
        self.name             = name
        self._cash            = initial_capital
        self._initial_capital = initial_capital
        self._signal          = signal
        self._risk            = risk_engine
        
        self._positions    : dict[str, Position] = {}
        self._trade_log    : list[dict]          = []
        self._equity_curve : list[float]         = [initial_capital]
        self._price_data   : dict[str, list]     = {}
    
    # Properties
    @property
    def cash(self) -> float:
        return self._cash
    
    @property
    def positions_value(self) -> float:
        return sum(p.market_value for p in self._positions.values())
    
    @property
    def total_value(self) -> float:
        return self._cash + self.positions_value
    
    @property
    def total_return(self) -> float:
        return (self.total_value - self._initial_capital) / self._initial_capital
    
    @property
    def current_drawdown(self) -> float:
        return self._risk.current_drawdown(self._equity_curve)
    
    # Dunder methods
    def __len__(self) -> int:
        return len(self._positions)
    
    def __contains__(self, ticker: str) -> bool:
        return ticker in self._positions
    
    def __repr__(self) -> str:
        return (f"PortfolioEngine('{self.name}', "
                f"NAV=${self.total_value:,.0f}, "
                f"positions={len(self)}, "
                f"return={self.total_return:+.2%})")
    
    # Core Methods
    def update_price(self, ticker: str, price: float) -> None:
        if ticker in self._positions:
            self._positions[ticker].current_price = price
        if ticker not in self._price_data:
            self._price_data[ticker] = []
        self._price_data[ticker].append(price)
    
    def _execute_trade(self, ticker: str, asset: Asset, 
                       shares: float, price: float, side: str) -> bool:
        order_value = shares * price
        
        if not self._risk.check_position_limit(order_value, self.total_value):
            print(f"  ⚠️  Position limit breached for {ticker}")
            return False
        
        if not self._risk.check_drawdown_limit(self._equity_curve):
            print(f"  🛑 Drawdown limit — halting trading")
            return False
        
        if side == "BUY":
            if order_value > self._cash:
                shares = int(self._cash / price)
                order_value = shares * price
                print(f"  ⚠️  Partial fill: {shares} shares")
            
            if shares <= 0: return False
            self._cash -= order_value
            
            if ticker in self._positions:
                pos = self._positions[ticker]
                total_shares = pos.shares + shares
                avg = (pos.shares * pos.avg_cost + shares * price) / total_shares
                pos.shares, pos.avg_cost = total_shares, avg
            else:
                self._positions[ticker] = Position(asset, shares, price, price)
                
        elif side == "SELL":
            if ticker not in self._positions:
                print(f"  ❌ No position in {ticker}")
                return False
            pos = self._positions[ticker]
            shares = min(shares, pos.shares)
            self._cash += shares * price
            pos.shares -= shares
            if pos.shares <= 0:
                del self._positions[ticker]
        
        self._trade_log.append({
            "timestamp": datetime.now().isoformat(),
            "ticker": ticker, "side": side,
            "shares": shares, "price": price,
            "value": shares * price
        })
        return True
    
    def run_signal(self, ticker: str, asset: Asset, target_weight: float = 0.15) -> None:
        if ticker not in self._price_data or len(self._price_data[ticker]) < 2:
            return
        
        prices_series = pd.Series(self._price_data[ticker])
        sig = self._signal.compute(prices_series)
        current_price = self._price_data[ticker][-1]
        
        target_value   = sig * target_weight * self.total_value
        current_value  = self._positions.get(ticker, Position(asset, 0, 0, current_price)).market_value
        diff_value     = target_value - current_value
        diff_shares    = int(diff_value / current_price)
        
        if abs(diff_shares) > 0:
            side = "BUY" if diff_shares > 0 else "SELL"
            self._execute_trade(ticker, asset, abs(diff_shares), current_price, side)
    
    def mark_to_market(self) -> None:
        self._equity_curve.append(self.total_value)
    
    def performance_report(self) -> pd.DataFrame:
        ec = np.array(self._equity_curve)
        daily_rets = np.diff(ec) / ec[:-1]
        
        if len(daily_rets) < 2:
            return pd.DataFrame()
        
        ann_ret  = (ec[-1]/ec[0]) ** (252/len(daily_rets)) - 1
        ann_vol  = daily_rets.std(ddof=1) * np.sqrt(252)
        sharpe   = (ann_ret - 0.05) / ann_vol if ann_vol > 0 else 0
        
        rolling_max = np.maximum.accumulate(ec)
        drawdowns   = (rolling_max - ec) / rolling_max
        max_dd      = drawdowns.max()
        calmar      = ann_ret / max_dd if max_dd > 0 else float('inf')
        
        var95  = self._risk.historical_var(list(daily_rets))
        cvar95 = self._risk.historical_cvar(list(daily_rets))
        
        metrics = {
            "Total Return"     : f"{self.total_return:+.2%}",
            "Annual Return"    : f"{ann_ret:+.2%}",
            "Annual Volatility": f"{ann_vol:.2%}",
            "Sharpe Ratio"     : f"{sharpe:.3f}",
            "Max Drawdown"     : f"{max_dd:.2%}",
            "Calmar Ratio"     : f"{calmar:.3f}",
            "95% VaR (daily)"  : f"{var95:.2%}",
            "95% CVaR (daily)" : f"{cvar95:.2%}",
            "Total Trades"     : len(self._trade_log),
            "Open Positions"   : len(self._positions),
            "Current NAV"      : f"${self.total_value:,.2f}",
        }
        return pd.DataFrame.from_dict(metrics, orient="index", columns=[self.name])
    
    def positions_summary(self) -> pd.DataFrame:
        if not self._positions:
            return pd.DataFrame()
        rows = []
        for ticker, pos in self._positions.items():
            rows.append({
                "Ticker"        : ticker,
                "Sector"        : pos.asset.sector,
                "Shares"        : pos.shares,
                "Avg Cost"      : f"${pos.avg_cost:.2f}",
                "Current Price" : f"${pos.current_price:.2f}",
                "Market Value"  : f"${pos.market_value:,.0f}",
                "Unrealized PnL": f"${pos.unrealized_pnl:+,.0f}",
                "PnL %"         : f"{pos.pnl_pct:+.2%}",
                "Weight"        : f"{pos.market_value/self.total_value:.1%}",
            })
        return pd.DataFrame(rows).set_index("Ticker")


print("✅ Portfolio Engine classes defined successfully.")
print(f"   Version: {PortfolioEngine.VERSION}\n")
```

```python
# ═══════════════════════════════════════════════════════════════════
# RUN SIMULATION
# ═══════════════════════════════════════════════════════════════════

universe = [
    Asset("AAPL", "Apple Inc.", "Technology"),
    Asset("MSFT", "Microsoft Corp.", "Technology"),
    Asset("GOOGL", "Alphabet Inc.", "Technology"),
    Asset("JPM", "JP Morgan Chase", "Financials"),
    Asset("GS", "Goldman Sachs", "Financials"),
    Asset("JNJ", "Johnson & Johnson", "Healthcare"),
    Asset("XOM", "Exxon Mobil", "Energy"),
]

initial_prices = {"AAPL":150, "MSFT":380, "GOOGL":140, "JPM":150, "GS":330, "JNJ":160, "XOM":105}
mus = {"AAPL":0.15, "MSFT":0.20, "GOOGL":0.12, "JPM":0.10, "GS":0.14, "JNJ":0.08, "XOM":0.07}
sigmas = {"AAPL":0.20, "MSFT":0.22, "GOOGL":0.24, "JPM":0.25, "GS":0.28, "JNJ":0.15, "XOM":0.22}

signal = CompositeSignal([(MomentumSignal(10), 0.6), (MeanReversionSignal(15), 0.4)])
risk = RiskEngine(0.20, 0.15, 0.95)

engine = PortfolioEngine("AlphaQuant Fund I", 1_000_000, signal, risk)

print(f"Portfolio Engine: {repr(engine)}")
print(f"Signal: {signal.name}\n")

print("Running 252-day simulation...".center(60, "="))

for day in range(252):
    for asset in universe:
        ticker = asset.ticker
        prev = engine._price_data.get(ticker, [initial_prices[ticker]])[-1]
        shock = np.random.normal(mus[ticker]/252, sigmas[ticker]/np.sqrt(252))
        new_p = prev * np.exp(shock)
        engine.update_price(ticker, new_p)

    for asset in universe:
        engine.run_signal(asset.ticker, asset, target_weight=0.12)

    engine.mark_to_market()

    if (day + 1) % 60 == 0 or day == 251:
        print(f"Day {day+1:3d} | NAV ${engine.total_value:,.0f} | "
              f"Return {engine.total_return:+.2%} | DD {engine.current_drawdown:.2%} | "
              f"Positions {len(engine)}")

print("\n" + "="*60)
print("PERFORMANCE REPORT")
print("="*60)
print(engine.performance_report())

print("\nOPEN POSITIONS")
print(engine.positions_summary())

print("\n✅ Simulation Completed Successfully!")
```
