## 2. Data Collection

We fetch OHLCV data for Gold (GC=F) and Silver (SI=F) via `yfinance`. These are the front-month continuous futures proxies available on Yahoo Finance. **Important caveat:** yfinance futures data does not include roll adjustments — in a production system, you would use back-adjusted continuous contracts from a provider such as Quandl/Nasdaq Data Link or Bloomberg.

```python
# ============================================================
# CELL 2: Data Loader
# ============================================================

def data_loader(
    tickers: list,
    start: str = '2010-01-01',
    end: str   = None,
    interval: str = '1d'
) -> dict:
    """
    Fetch OHLCV data from yfinance for a list of tickers.

    Returns
    -------
    dict: {ticker: pd.DataFrame} with clean OHLCV columns.
    """
    if end is None:
        end = datetime.today().strftime('%Y-%m-%d')

    data = {}
    for ticker in tickers:
        print(f"  Fetching {ticker} ...", end=' ')
        raw = yf.download(ticker, start=start, end=end, interval=interval,
                          progress=False, auto_adjust=True)

        if raw.empty:
            print(f"WARNING: No data for {ticker}")
            continue

        # Flatten MultiIndex columns if present
        if isinstance(raw.columns, pd.MultiIndex):
            raw.columns = raw.columns.get_level_values(0)

        # Standardise column names
        raw.columns = [c.lower() for c in raw.columns]
        required = ['open', 'high', 'low', 'close', 'volume']
        raw = raw[[c for c in required if c in raw.columns]]

        # Drop rows where close is NaN or zero
        raw = raw[raw['close'].notna() & (raw['close'] > 0)]

        # Forward-fill remaining gaps (weekends already excluded for daily)
        raw = raw.ffill().dropna()

        data[ticker] = raw
        print(f"✅  {len(raw)} rows  [{raw.index[0].date()} → {raw.index[-1].date()}]")

    return data


# ── Fetch data ────────────────────────────────────────────────
print("📥 Downloading bullion futures data...")
raw_data = data_loader(
    tickers=['GC=F', 'SI=F'],
    start='2010-01-01'
)

gc = raw_data['GC=F'].copy()
si = raw_data['SI=F'].copy()

# ── Align on common dates ─────────────────────────────────────
common_idx = gc.index.intersection(si.index)
gc = gc.loc[common_idx]
si = si.loc[common_idx]

print(f"\n📊 Aligned dataset: {len(common_idx)} trading days")
print(f"   Range: {common_idx[0].date()} → {common_idx[-1].date()}")

# ── Build combined close DataFrame ────────────────────────────
prices = pd.DataFrame({
    'GC': gc['close'],
    'SI': si['close']
}, index=common_idx)

print("\n📋 Price snapshot (last 5 rows):")
print(prices.tail())
```

## 3. Feature Engineering

We construct a rich set of signals across four dimensions:
- **Trend/Momentum:** Moving average crossovers, Donchian channel breakouts
- **Mean Reversion:** Z-score of returns, Bollinger Bands
- **Volatility:** Rolling realized volatility, regime classification
- **Cross-Asset:** Gold/Silver ratio, relative strength

```python
# ============================================================
# CELL 4: Feature Engineering
# ============================================================

def signal_generator(
    df: pd.DataFrame,
    short_win: int = 20,
    long_win: int  = 60,
    bb_win: int    = 20,
    bb_std: float  = 2.0,
    don_win: int   = 20,
    vol_win: int   = 20,
    zscore_win: int = 20
) -> pd.DataFrame:
    """
    Generate technical features for a single instrument OHLCV DataFrame.

    Parameters
    ----------
    df         : OHLCV DataFrame with columns [open, high, low, close, volume]
    short_win  : Short MA window
    long_win   : Long MA window
    bb_win     : Bollinger Band window
    bb_std     : Bollinger Band standard deviation multiplier
    don_win    : Donchian channel window
    vol_win    : Rolling volatility window
    zscore_win : Z-score of returns window

    Returns
    -------
    pd.DataFrame with all features appended
    """
    f = df.copy()
    c = f['close']

    # ── Daily log returns ──────────────────────────────────────
    f['ret']       = np.log(c / c.shift(1))
    f['ret_5d']    = np.log(c / c.shift(5))
    f['ret_20d']   = np.log(c / c.shift(20))

    # ── Trend: Moving Averages ─────────────────────────────────
    f['ma_short']  = c.rolling(short_win).mean()
    f['ma_long']   = c.rolling(long_win).mean()
    f['ma_signal'] = np.where(f['ma_short'] > f['ma_long'], 1.0, -1.0)
    f['ma_ratio']  = f['ma_short'] / f['ma_long'] - 1   # normalized spread

    # ── Trend: Donchian Channel ────────────────────────────────
    f['don_high']  = f['high'].rolling(don_win).max()
    f['don_low']   = f['low'].rolling(don_win).min()
    f['don_mid']   = (f['don_high'] + f['don_low']) / 2
    f['don_width'] = (f['don_high'] - f['don_low']) / f['don_mid']
    # Breakout signal: +1 if price breaks above channel high, -1 if below low
    f['don_signal'] = 0.0
    f.loc[c >= f['don_high'].shift(1), 'don_signal'] =  1.0
    f.loc[c <= f['don_low'].shift(1),  'don_signal'] = -1.0

    # ── Mean Reversion: Bollinger Bands ────────────────────────
    f['bb_mid']    = c.rolling(bb_win).mean()
    bb_std_roll    = c.rolling(bb_win).std()
    f['bb_upper']  = f['bb_mid'] + bb_std * bb_std_roll
    f['bb_lower']  = f['bb_mid'] - bb_std * bb_std_roll
    f['bb_pct']    = (c - f['bb_lower']) / (f['bb_upper'] - f['bb_lower'])
    # Mean reversion: sell when > upper, buy when < lower
    f['bb_signal'] = 0.0
    f.loc[c > f['bb_upper'], 'bb_signal'] = -1.0
    f.loc[c < f['bb_lower'], 'bb_signal'] =  1.0

    # ── Mean Reversion: Z-score of returns ─────────────────────
    ret_mean       = f['ret'].rolling(zscore_win).mean()
    ret_std        = f['ret'].rolling(zscore_win).std()
    f['ret_zscore']= (f['ret'] - ret_mean) / ret_std
    # Fade extreme moves: buy when very negative, sell when very positive
    f['zscore_signal'] = -np.sign(f['ret_zscore'])  # mean-revert

    # ── Volatility ─────────────────────────────────────────────
    f['rvol']      = f['ret'].rolling(vol_win).std() * np.sqrt(ANN_FACTOR)
    f['rvol_ma']   = f['rvol'].rolling(vol_win).mean()
    # Volatility regime: 1 = high vol, 0 = low vol
    f['vol_regime']= (f['rvol'] > f['rvol_ma']).astype(float)

    return f


def compute_cross_asset_signals(gc_df: pd.DataFrame, si_df: pd.DataFrame,
                                 ratio_win: int = 20) -> pd.DataFrame:
    """
    Compute Gold/Silver ratio and relative strength signals.
    """
    ca = pd.DataFrame(index=gc_df.index)
    ca['gc_close']    = gc_df['close']
    ca['si_close']    = si_df['close']
    ca['ratio']       = gc_df['close'] / si_df['close']

    # Z-score of ratio for pairs trade
    ratio_mean        = ca['ratio'].rolling(ratio_win).mean()
    ratio_std         = ca['ratio'].rolling(ratio_win).std()
    ca['ratio_zscore']= (ca['ratio'] - ratio_mean) / ratio_std

    # Relative strength: 20-day return diff
    ca['gc_ret20']    = np.log(ca['gc_close'] / ca['gc_close'].shift(20))
    ca['si_ret20']    = np.log(ca['si_close'] / ca['si_close'].shift(20))
    ca['rel_strength']= ca['gc_ret20'] - ca['si_ret20']  # +ve: gold outperforms

    return ca


# ── Generate signals ──────────────────────────────────────────
print("⚙️  Generating features for Gold...")
gc_feat = signal_generator(gc, short_win=20, long_win=60)

print("⚙️  Generating features for Silver...")
si_feat = signal_generator(si, short_win=20, long_win=60)

print("⚙️  Computing cross-asset signals...")
cross = compute_cross_asset_signals(gc, si, ratio_win=40)

# Drop rows with NaN from rolling calculations
start_idx = max(60, 40)  # warmup period
gc_feat = gc_feat.iloc[start_idx:]
si_feat = si_feat.iloc[start_idx:]
cross   = cross.iloc[start_idx:]

print(f"\n✅ Features computed. Usable rows: {len(gc_feat)}")
print("\nGold feature columns:", list(gc_feat.columns))
```

## 4. Strategy Design & Backtesting Engine

### Strategy A — Trend Following (Dual MA Crossover)

**Entry/Exit Rules:**
- **Long:** MA(20) crosses above MA(60) → position = +1
- **Short:** MA(20) crosses below MA(60) → position = −1
- Applied independently to both Gold and Silver

**Position Sizing:** Volatility-scaled (target 10% ann. vol), capped at 3 contracts

**Risk Management:** Hard stop at 2% of NAV per instrument

---

### Strategy B — Relative Value (Pairs Trade)

**Entry/Exit Rules:**
- **Long Gold / Short Silver:** Ratio Z-score > +1.5 (ratio elevated → expect mean reversion down → short gold, long silver... wait: if ratio high, gold expensive vs silver → sell gold, buy silver. Or equivalently, if ratio Z-score > 1.5, go SHORT ratio: short GC + long SI)
- **Long Silver / Short Gold:** Z-score < −1.5 (ratio depressed → buy ratio: long GC + short SI)
- **Exit:** |Z-score| < 0.5

**Position Sizing:** Equal dollar notional on both legs (dollar-neutral), volatility-adjusted

```python
# ============================================================
# CELL 6: Backtesting Engine Core
# ============================================================

def compute_position_size(
    raw_signal: pd.Series,
    rvol: pd.Series,
    price: pd.Series,
    capital: float = CAPITAL,
    target_vol: float = TARGET_VOL,
    max_leverage: float = 3.0
) -> pd.Series:
    """
    Volatility-scaled position sizing.

    Dollar position = (target_vol / realized_vol) * capital * signal_direction
    Clamped to [-max_leverage * capital, +max_leverage * capital].
    """
    # rvol guard: avoid division by zero
    rvol_safe = rvol.clip(lower=0.01)

    # Dollar notional target
    dollar_pos = (target_vol / rvol_safe) * capital * raw_signal

    # Cap leverage
    max_pos = max_leverage * capital
    dollar_pos = dollar_pos.clip(-max_pos, max_pos)

    return dollar_pos  # in USD


def backtest_engine(
    signal: pd.Series,          # direction: +1, 0, -1
    price: pd.Series,           # close prices
    rvol: pd.Series,            # annualised realised vol
    instrument_name: str = 'X',
    capital: float = CAPITAL,
    target_vol: float = TARGET_VOL,
    cost_bps: float = TOTAL_COST_BPS,
    commission_usd: float = COMMISSION_USD,
    multiplier: float = 100.0
) -> pd.DataFrame:
    """
    Core backtesting engine. Returns daily PnL and position DataFrame.

    Assumptions
    -----------
    - Positions sized in USD notional (converted to contracts for commission)
    - Entry at next day's open (signal is end-of-day)
    - Mark-to-market PnL daily
    - Costs applied on position changes (trades)
    """
    sig  = signal.reindex(price.index).ffill().fillna(0)
    rvol = rvol.reindex(price.index).ffill().fillna(0.15)

    # Volatility-scaled USD position
    usd_pos = compute_position_size(sig, rvol, price, capital, target_vol)

    # Shift by 1: trade enters at next day's open (we use close as proxy)
    usd_pos_exec = usd_pos.shift(1).fillna(0)

    # Daily log returns of the instrument
    log_ret = np.log(price / price.shift(1)).fillna(0)

    # Gross PnL: position (USD) * return
    gross_pnl = usd_pos_exec * log_ret

    # ── Transaction Costs ──────────────────────────────────────
    pos_change    = usd_pos_exec.diff().abs()  # USD notional change
    spread_cost   = pos_change * (cost_bps / 10_000)      # spread + slippage

    # Commission: estimated number of contracts turned over
    n_contracts   = (pos_change / (price * multiplier)).fillna(0)
    commission    = n_contracts * commission_usd

    total_cost    = spread_cost + commission
    net_pnl       = gross_pnl - total_cost

    result = pd.DataFrame({
        f'{instrument_name}_signal':    sig,
        f'{instrument_name}_usd_pos':   usd_pos_exec,
        f'{instrument_name}_gross_pnl': gross_pnl,
        f'{instrument_name}_cost':      total_cost,
        f'{instrument_name}_net_pnl':   net_pnl,
    }, index=price.index)

    return result


print("✅ Backtesting engine defined.")
```

```python
# ============================================================
# CELL 7: Strategy A — Trend Following
# ============================================================

print("📈 Running Strategy A: Trend Following...")

# Align all series to the same index
idx = gc_feat.index.intersection(si_feat.index)
gc_feat = gc_feat.loc[idx]
si_feat = si_feat.loc[idx]
cross   = cross.loc[idx]

# ── Gold Trend ─────────────────────────────────────────────
gc_bt_a = backtest_engine(
    signal          = gc_feat['ma_signal'],
    price           = gc_feat['close'],
    rvol            = gc_feat['rvol'],
    instrument_name = 'GC_A',
    capital         = CAPITAL / 2,      # allocate 50% to each leg
    target_vol      = TARGET_VOL,
    multiplier      = GC_MULTIPLIER
)

# ── Silver Trend ────────────────────────────────────────────
si_bt_a = backtest_engine(
    signal          = si_feat['ma_signal'],
    price           = si_feat['close'],
    rvol            = si_feat['rvol'],
    instrument_name = 'SI_A',
    capital         = CAPITAL / 2,
    target_vol      = TARGET_VOL,
    multiplier      = SI_MULTIPLIER
)

# ── Combine Strategy A ──────────────────────────────────────
strat_a = pd.concat([gc_bt_a, si_bt_a], axis=1)
strat_a['A_net_pnl']   = strat_a['GC_A_net_pnl']   + strat_a['SI_A_net_pnl']
strat_a['A_gross_pnl'] = strat_a['GC_A_gross_pnl'] + strat_a['SI_A_gross_pnl']
strat_a['A_cost']      = strat_a['GC_A_cost']      + strat_a['SI_A_cost']

print(f"  GC Trend trades: {(gc_bt_a['GC_A_signal'].diff() != 0).sum()}")
print(f"  SI Trend trades: {(si_bt_a['SI_A_signal'].diff() != 0).sum()}")
print(f"  Total Strategy A gross PnL: ${strat_a['A_gross_pnl'].sum():,.0f}")
print(f"  Total Strategy A costs:     ${strat_a['A_cost'].sum():,.0f}")
print(f"  Total Strategy A net PnL:   ${strat_a['A_net_pnl'].sum():,.0f}")
```

```python
# ============================================================
# CELL 8: Strategy B — Relative Value (Pairs Trade)
# ============================================================

print("⚖️  Running Strategy B: Relative Value (Pairs Trade)...")

ENTRY_THRESH = 1.5   # Z-score threshold for entry
EXIT_THRESH  = 0.5   # Z-score threshold for exit

def build_pairs_signals(ratio_zscore: pd.Series,
                         entry: float = 1.5,
                         exit_: float = 0.5) -> pd.DataFrame:
    """
    Build GC/SI pairs trade signals from ratio Z-score.

    Logic:
      - Z > +entry → ratio too HIGH → SHORT ratio → SHORT GC, LONG SI
      - Z < -entry → ratio too LOW  → LONG  ratio → LONG  GC, SHORT SI
      - |Z| < exit → flat
    """
    gc_sig = pd.Series(0.0, index=ratio_zscore.index)
    si_sig = pd.Series(0.0, index=ratio_zscore.index)
    position = 0

    for dt in ratio_zscore.index:
        z = ratio_zscore.loc[dt]
        if np.isnan(z):
            continue

        if position == 0:
            if z > entry:       # Short GC, Long SI
                position = -1
            elif z < -entry:    # Long GC, Short SI
                position = 1
        elif position == 1:
            if z >= -exit_:     # Ratio recovered, exit
                position = 0
        elif position == -1:
            if z <= exit_:      # Ratio recovered, exit
                position = 0

        gc_sig.loc[dt] = float(position)     # +1 = long GC, -1 = short GC
        si_sig.loc[dt] = float(-position)    # opposite leg

    return pd.DataFrame({'gc_sig_b': gc_sig, 'si_sig_b': si_sig})


# Build pairs signals
pairs_sigs = build_pairs_signals(cross['ratio_zscore'], ENTRY_THRESH, EXIT_THRESH)

# Backtest each leg
gc_bt_b = backtest_engine(
    signal          = pairs_sigs['gc_sig_b'],
    price           = gc_feat['close'],
    rvol            = gc_feat['rvol'],
    instrument_name = 'GC_B',
    capital         = CAPITAL / 2,
    target_vol      = TARGET_VOL,
    multiplier      = GC_MULTIPLIER
)

si_bt_b = backtest_engine(
    signal          = pairs_sigs['si_sig_b'],
    price           = si_feat['close'],
    rvol            = si_feat['rvol'],
    instrument_name = 'SI_B',
    capital         = CAPITAL / 2,
    target_vol      = TARGET_VOL,
    multiplier      = SI_MULTIPLIER
)

# Combine
strat_b = pd.concat([gc_bt_b, si_bt_b, pairs_sigs], axis=1)
strat_b['B_net_pnl']   = strat_b['GC_B_net_pnl']   + strat_b['SI_B_net_pnl']
strat_b['B_gross_pnl'] = strat_b['GC_B_gross_pnl'] + strat_b['SI_B_gross_pnl']
strat_b['B_cost']      = strat_b['GC_B_cost']      + strat_b['SI_B_cost']

n_trades_b = (pairs_sigs['gc_sig_b'].diff() != 0).sum()
print(f"  Pairs trades (round-trips): {n_trades_b}")
print(f"  Total Strategy B gross PnL: ${strat_b['B_gross_pnl'].sum():,.0f}")
print(f"  Total Strategy B costs:     ${strat_b['B_cost'].sum():,.0f}")
print(f"  Total Strategy B net PnL:   ${strat_b['B_net_pnl'].sum():,.0f}")
```

```python
# ============================================================
# CELL 9: Portfolio Aggregation + Benchmarks
# ============================================================

print("🗂  Building combined portfolio and benchmarks...")

# ── Combined portfolio (50/50 A + B) ──────────────────────────
portfolio = pd.DataFrame(index=idx)
portfolio['A_net_pnl']   = strat_a['A_net_pnl']
portfolio['B_net_pnl']   = strat_b['B_net_pnl']
portfolio['port_net_pnl']= portfolio['A_net_pnl'] * 0.5 + portfolio['B_net_pnl'] * 0.5

# Gross versions for cost breakdown
portfolio['A_gross_pnl']     = strat_a['A_gross_pnl']
portfolio['B_gross_pnl']     = strat_b['B_gross_pnl']
portfolio['port_gross_pnl']  = portfolio['A_gross_pnl'] * 0.5 + portfolio['B_gross_pnl'] * 0.5
portfolio['total_cost']      = strat_a['A_cost'] * 0.5 + strat_b['B_cost'] * 0.5

# ── Cumulative NAV curves ──────────────────────────────────────
def pnl_to_nav(pnl_series: pd.Series, capital: float = CAPITAL) -> pd.Series:
    """Convert daily PnL series to NAV curve starting at capital."""
    return capital + pnl_series.cumsum()

portfolio['NAV_A']      = pnl_to_nav(portfolio['A_net_pnl'])
portfolio['NAV_B']      = pnl_to_nav(portfolio['B_net_pnl'])
portfolio['NAV_port']   = pnl_to_nav(portfolio['port_net_pnl'])
portfolio['NAV_A_gross']= pnl_to_nav(portfolio['A_gross_pnl'])

# ── Buy-and-hold benchmarks ────────────────────────────────────
gc_bh_ret = np.log(gc_feat['close'] / gc_feat['close'].shift(1)).fillna(0)
si_bh_ret = np.log(si_feat['close'] / si_feat['close'].shift(1)).fillna(0)

portfolio['BH_GC_pnl']  = CAPITAL * gc_bh_ret
portfolio['BH_SI_pnl']  = CAPITAL * si_bh_ret
portfolio['NAV_BH_GC']  = pnl_to_nav(portfolio['BH_GC_pnl'])
portfolio['NAV_BH_SI']  = pnl_to_nav(portfolio['BH_SI_pnl'])

print(f"✅ Portfolio NAV range: ${portfolio['NAV_port'].min():,.0f} – ${portfolio['NAV_port'].max():,.0f}")
print(f"   Final NAV (Strategy A):   ${portfolio['NAV_A'].iloc[-1]:,.0f}")
print(f"   Final NAV (Strategy B):   ${portfolio['NAV_B'].iloc[-1]:,.0f}")
print(f"   Final NAV (Combined):     ${portfolio['NAV_port'].iloc[-1]:,.0f}")
print(f"   Final NAV (B&H Gold):     ${portfolio['NAV_BH_GC'].iloc[-1]:,.0f}")
```

## 5. Performance Evaluation

We compute a comprehensive set of risk-adjusted performance metrics for each strategy and the combined portfolio.

```python
# ============================================================
# CELL 10: Performance Metrics
# ============================================================

def performance_metrics(
    pnl: pd.Series,
    capital: float = CAPITAL,
    ann: int = ANN_FACTOR,
    label: str = 'Strategy'
) -> dict:
    """
    Compute comprehensive performance metrics from daily PnL series.

    Returns
    -------
    dict with: Sharpe, Sortino, MaxDD, AnnReturn, AnnVol, HitRate, etc.
    """
    pnl = pnl.dropna()
    if len(pnl) == 0:
        return {}

    nav   = capital + pnl.cumsum()
    rets  = pnl / capital                # daily returns as fraction of initial capital

    # Annualised return
    total_ret  = (nav.iloc[-1] / nav.iloc[0]) - 1
    n_years    = len(pnl) / ann
    ann_ret    = (1 + total_ret) ** (1 / n_years) - 1 if n_years > 0 else 0

    # Annualised volatility
    ann_vol    = rets.std() * np.sqrt(ann)

    # Sharpe ratio (zero risk-free rate for simplicity)
    sharpe     = (rets.mean() / rets.std()) * np.sqrt(ann) if rets.std() > 0 else 0

    # Sortino ratio (downside deviation)
    downside   = rets[rets < 0]
    sortino    = (rets.mean() / downside.std()) * np.sqrt(ann) if len(downside) > 0 else 0

    # Maximum drawdown
    rolling_max = nav.cummax()
    drawdown    = (nav - rolling_max) / rolling_max
    max_dd      = drawdown.min()

    # Calmar ratio
    calmar      = ann_ret / abs(max_dd) if max_dd != 0 else 0

    # Hit rate
    hit_rate    = (pnl > 0).mean()

    # Average win vs loss
    avg_win     = pnl[pnl > 0].mean() if (pnl > 0).any() else 0
    avg_loss    = pnl[pnl < 0].mean() if (pnl < 0).any() else 0
    profit_factor = abs(avg_win / avg_loss) if avg_loss != 0 else np.inf

    # Total return
    total_pnl   = pnl.sum()

    return {
        'Label':           label,
        'Total PnL ($)':   f'{total_pnl:,.0f}',
        'Total Return (%)': f'{total_ret*100:.1f}%',
        'Ann. Return (%)': f'{ann_ret*100:.1f}%',
        'Ann. Vol (%)':    f'{ann_vol*100:.1f}%',
        'Sharpe':          f'{sharpe:.2f}',
        'Sortino':         f'{sortino:.2f}',
        'Calmar':          f'{calmar:.2f}',
        'Max DD (%)':      f'{max_dd*100:.1f}%',
        'Hit Rate (%)':    f'{hit_rate*100:.1f}%',
        'Profit Factor':   f'{profit_factor:.2f}',
        'Avg Win ($)':     f'{avg_win:,.0f}',
        'Avg Loss ($)':    f'{avg_loss:,.0f}',
    }


# ── Compute metrics for all strategies ────────────────────────
metrics = [
    performance_metrics(portfolio['A_net_pnl'],    label='Strategy A (Trend)'),
    performance_metrics(portfolio['A_gross_pnl'],  label='Strategy A (Gross)'),
    performance_metrics(portfolio['B_net_pnl'],    label='Strategy B (Pairs)'),
    performance_metrics(portfolio['port_net_pnl'], label='Combined Portfolio'),
    performance_metrics(portfolio['BH_GC_pnl'],    label='B&H Gold'),
    performance_metrics(portfolio['BH_SI_pnl'],    label='B&H Silver'),
]

metrics_df = pd.DataFrame(metrics).set_index('Label')
print("\n🏆 Performance Summary")
print("=" * 90)
print(metrics_df.to_string())
```

```python
# ============================================================
# CELL 13: Rolling Sharpe Ratio
# ============================================================

def rolling_sharpe(pnl: pd.Series, window: int = 60, capital: float = CAPITAL) -> pd.Series:
    """Compute rolling annualised Sharpe ratio."""
    rets = pnl / capital
    roll_mean = rets.rolling(window).mean()
    roll_std  = rets.rolling(window).std()
    return (roll_mean / roll_std) * np.sqrt(ANN_FACTOR)


def plot_rolling_sharpe(portfolio: pd.DataFrame, window: int = 60) -> go.Figure:
    fig = go.Figure()

    series_map = {
        'A_net_pnl':    ('Strategy A (Trend)', '#1ABC9C'),
        'B_net_pnl':    ('Strategy B (Pairs)', '#3498DB'),
        'port_net_pnl': ('Combined',           '#FFD700'),
        'BH_GC_pnl':    ('B&H Gold',           '#E74C3C'),
    }

    for col, (label, color) in series_map.items():
        rs = rolling_sharpe(portfolio[col], window)
        fig.add_trace(go.Scatter(
            x=rs.index, y=rs.values,
            name=label, line=dict(color=color, width=1.5)
        ))

    fig.add_hline(y=0,  line_dash='dash', line_color='white', opacity=0.4)
    fig.add_hline(y=1,  line_dash='dot',  line_color='#2ECC71', opacity=0.6,
                  annotation_text='Sharpe = 1', annotation_position='right')
    fig.add_hline(y=-1, line_dash='dot',  line_color='#E74C3C', opacity=0.4)

    fig.update_layout(
        height=400, template='plotly_dark',
        title=dict(text=f'📈 Rolling {window}-Day Sharpe Ratio', font=dict(size=18)),
        yaxis_title='Sharpe Ratio', xaxis_title='Date',
        hovermode='x unified'
    )
    return fig


fig_sharpe = plot_rolling_sharpe(portfolio, window=60)
fig_sharpe.show()
```

```python
# ============================================================
# CELL 14: Signal Overlay on Gold Price Chart
# ============================================================

def plot_signal_overlay(feat_df: pd.DataFrame, bt_df: pd.DataFrame,
                         signal_col: str, label: str = 'Gold') -> go.Figure:
    """Plot price with buy/sell signals overlaid."""

    sig   = bt_df[signal_col].reindex(feat_df.index).ffill().fillna(0)
    price = feat_df['close']

    # Detect signal changes (trade events)
    sig_change  = sig.diff().fillna(0)
    buy_dates   = price[sig_change > 0].index
    sell_dates  = price[sig_change < 0].index

    # Show only last 3 years for clarity
    cutoff = feat_df.index[-1] - pd.DateOffset(years=3)
    feat_sub = feat_df.loc[feat_df.index >= cutoff]
    buy_sub  = [d for d in buy_dates  if d >= cutoff]
    sell_sub = [d for d in sell_dates if d >= cutoff]

    fig = make_subplots(
        rows=2, cols=1, shared_xaxes=True,
        row_heights=[0.70, 0.30],
        subplot_titles=[f'{label} — Price with Trade Signals (Last 3Y)',
                        'Signal Position (+1 Long / -1 Short)'],
        vertical_spacing=0.06
    )

    # Price
    fig.add_trace(go.Scatter(
        x=feat_sub.index, y=feat_sub['close'],
        name=f'{label} Close', line=dict(color='#FFD700', width=1.5)
    ), row=1, col=1)

    # MA overlays
    fig.add_trace(go.Scatter(
        x=feat_sub.index, y=feat_sub['ma_short'],
        name='MA(20)', line=dict(color='#1ABC9C', width=1, dash='dash')
    ), row=1, col=1)
    fig.add_trace(go.Scatter(
        x=feat_sub.index, y=feat_sub['ma_long'],
        name='MA(60)', line=dict(color='#E74C3C', width=1, dash='dash')
    ), row=1, col=1)

    # Buy signals
    if buy_sub:
        fig.add_trace(go.Scatter(
            x=buy_sub, y=feat_sub.loc[buy_sub, 'close'],
            mode='markers', name='Buy / Cover Short',
            marker=dict(symbol='triangle-up', size=10, color='#2ECC71')
        ), row=1, col=1)

    # Sell signals
    if sell_sub:
        fig.add_trace(go.Scatter(
            x=sell_sub, y=feat_sub.loc[sell_sub, 'close'],
            mode='markers', name='Sell / Go Short',
            marker=dict(symbol='triangle-down', size=10, color='#E74C3C')
        ), row=1, col=1)

    # Position bar
    sig_sub = sig.loc[feat_sub.index]
    fig.add_trace(go.Bar(
        x=sig_sub.index, y=sig_sub.values,
        name='Signal', marker_color=np.where(sig_sub >= 0, '#2ECC71', '#E74C3C')
    ), row=2, col=1)

    fig.update_layout(
        height=600, template='plotly_dark',
        title=dict(text=f'🎯 {label} — Trade Signal Overlay', font=dict(size=18))
    )
    fig.update_yaxes(title_text='Price (USD/oz)', row=1, col=1)
    fig.update_yaxes(title_text='Direction', row=2, col=1)
    return fig


fig_gc_signals = plot_signal_overlay(gc_feat, gc_bt_a, 'GC_A_signal', label='Gold (Strategy A)')
fig_gc_signals.show()
```

## 7. Sensitivity Analysis

We test robustness by varying the key parameters of each strategy and measuring the impact on Sharpe ratio.

```python
# ============================================================
# CELL 16: Sensitivity Analysis — Strategy A (MA Windows)
# ============================================================

print("🔬 Running sensitivity analysis for Strategy A...")

short_windows = [5, 10, 20, 30, 40]
long_windows  = [30, 50, 60, 90, 120]

sens_results = []

for sw, lw in product(short_windows, long_windows):
    if sw >= lw:
        continue

    # Recompute signals
    gc_f = signal_generator(gc, short_win=sw, long_win=lw)
    gc_f = gc_f.dropna()

    if len(gc_f) < 200:
        continue

    bt = backtest_engine(
        signal=gc_f['ma_signal'],
        price=gc_f['close'],
        rvol=gc_f['rvol'],
        instrument_name='GC',
        capital=CAPITAL,
        target_vol=TARGET_VOL,
        multiplier=GC_MULTIPLIER
    )

    rets = bt['GC_net_pnl'] / CAPITAL
    if rets.std() < 1e-10:
        continue

    sharpe = (rets.mean() / rets.std()) * np.sqrt(ANN_FACTOR)
    total_ret = bt['GC_net_pnl'].sum() / CAPITAL * 100

    # Max DD
    nav = CAPITAL + bt['GC_net_pnl'].cumsum()
    max_dd = ((nav - nav.cummax()) / nav.cummax()).min() * 100

    sens_results.append({
        'Short MA': sw, 'Long MA': lw,
        'Sharpe': round(sharpe, 3),
        'Total Ret (%)': round(total_ret, 1),
        'Max DD (%)': round(max_dd, 1)
    })

sens_df = pd.DataFrame(sens_results)
print(f"  Combinations tested: {len(sens_df)}")
print(f"  Best Sharpe: {sens_df['Sharpe'].max():.2f}  "
      f"(Short={sens_df.loc[sens_df['Sharpe'].idxmax(), 'Short MA']}, "
      f"Long={sens_df.loc[sens_df['Sharpe'].idxmax(), 'Long MA']})")
```

```python
# ============================================================
# CELL 17: Sensitivity Heatmap
# ============================================================

def plot_sensitivity_heatmap(sens_df: pd.DataFrame, metric: str = 'Sharpe') -> go.Figure:
    """Plot a heatmap of Sharpe ratio vs MA parameters."""

    pivot = sens_df.pivot(index='Short MA', columns='Long MA', values=metric)

    fig = go.Figure(data=go.Heatmap(
        z=pivot.values,
        x=[str(c) for c in pivot.columns],
        y=[str(r) for r in pivot.index],
        colorscale='RdYlGn',
        zmid=0,
        text=np.round(pivot.values, 2),
        texttemplate='%{text}',
        colorbar=dict(title=metric)
    ))

    fig.update_layout(
        height=420, template='plotly_dark',
        title=dict(text=f'🔬 Sensitivity Analysis — {metric} by MA Windows (Gold)', font=dict(size=18)),
        xaxis_title='Long MA Window',
        yaxis_title='Short MA Window'
    )
    return fig


fig_sens_sharpe = plot_sensitivity_heatmap(sens_df, 'Sharpe')
fig_sens_sharpe.show()

fig_sens_ret = plot_sensitivity_heatmap(sens_df, 'Total Ret (%)')
fig_sens_ret.show()
```

```python
# ============================================================
# CELL 18: Sensitivity Analysis — Strategy B (Z-score thresholds)
# ============================================================

print("🔬 Running sensitivity analysis for Strategy B (Z-score thresholds)...")

entry_thresholds = [0.5, 1.0, 1.5, 2.0, 2.5]
exit_thresholds  = [0.1, 0.25, 0.5, 0.75, 1.0]

sens_b_results = []

for entry, exit_ in product(entry_thresholds, exit_thresholds):
    if exit_ >= entry:
        continue

    ps = build_pairs_signals(cross['ratio_zscore'], entry=entry, exit_=exit_)

    gc_b = backtest_engine(
        signal=ps['gc_sig_b'], price=gc_feat['close'], rvol=gc_feat['rvol'],
        instrument_name='GC', capital=CAPITAL/2, target_vol=TARGET_VOL,
        multiplier=GC_MULTIPLIER
    )
    si_b = backtest_engine(
        signal=ps['si_sig_b'], price=si_feat['close'], rvol=si_feat['rvol'],
        instrument_name='SI', capital=CAPITAL/2, target_vol=TARGET_VOL,
        multiplier=SI_MULTIPLIER
    )

    combined_pnl = gc_b['GC_net_pnl'] + si_b['SI_net_pnl']
    rets = combined_pnl / CAPITAL
    if rets.std() < 1e-10:
        continue

    sharpe = (rets.mean() / rets.std()) * np.sqrt(ANN_FACTOR)
    total_ret = combined_pnl.sum() / CAPITAL * 100

    sens_b_results.append({
        'Entry Z': entry, 'Exit Z': exit_,
        'Sharpe': round(sharpe, 3),
        'Total Ret (%)': round(total_ret, 1)
    })

sens_b_df = pd.DataFrame(sens_b_results)
print(f"  Combinations tested: {len(sens_b_df)}")
print(f"  Best Sharpe: {sens_b_df['Sharpe'].max():.2f}")

fig_sens_b = plot_sensitivity_heatmap(sens_b_df.rename(columns={'Entry Z':'Short MA','Exit Z':'Long MA'}), 'Sharpe')
fig_sens_b.update_layout(
    title='🔬 Strategy B Sensitivity — Sharpe by Entry/Exit Z-Score',
    xaxis_title='Exit Z-Score', yaxis_title='Entry Z-Score'
)
fig_sens_b.show()
```

```python
# ============================================================
# CELL 19: Volatility Regime Analysis
# ============================================================

print("📊 Analysing performance by volatility regime...")

portfolio['vol_regime'] = gc_feat['vol_regime']

def regime_metrics(pnl: pd.Series, regime: pd.Series, capital: float = CAPITAL) -> pd.DataFrame:
    """Split performance by volatility regime."""
    results = []
    for reg, label in [(0, 'Low Vol'), (1, 'High Vol')]:
        mask = regime == reg
        sub  = pnl[mask]
        if len(sub) < 10:
            continue
        rets = sub / capital
        sharpe = (rets.mean() / rets.std()) * np.sqrt(ANN_FACTOR) if rets.std() > 0 else 0
        ann_ret = rets.mean() * ANN_FACTOR
        hit = (sub > 0).mean()
        results.append({
            'Regime': label,
            'Days': int(mask.sum()),
            'Ann. Return (%)': f'{ann_ret*100:.1f}%',
            'Sharpe': f'{sharpe:.2f}',
            'Hit Rate': f'{hit*100:.1f}%'
        })
    return pd.DataFrame(results)

print("\n--- Strategy A by Volatility Regime ---")
print(regime_metrics(portfolio['A_net_pnl'], portfolio['vol_regime']))

print("\n--- Strategy B by Volatility Regime ---")
print(regime_metrics(portfolio['B_net_pnl'], portfolio['vol_regime']))

print("\n--- Combined Portfolio by Volatility Regime ---")
print(regime_metrics(portfolio['port_net_pnl'], portfolio['vol_regime']))
```

## 8. Interpretation & Insights

### Why Strategy A (Trend Following) Works in Bullion

Gold and silver are macroeconomic assets driven by:
- **Inflation regimes:** During inflationary periods, institutional flows into bullion are persistent and trend-like, creating multi-month momentum.
- **Safe haven demand:** Risk-off episodes create abrupt but sustained rallies (2008, 2020 COVID, 2022 Russia-Ukraine), which trend strategies capture well.
- **Futures market structure:** Managed money (CTAs, hedge funds) are large participants in gold/silver futures and are predominantly trend-following, creating self-reinforcing momentum.

**Why it can fail:**
- Choppy, range-bound markets (e.g. 2014–2018 sideways gold consolidation) generate repeated small losses from whipsaws.
- Volatility-scaling helps dampen losses during low-conviction periods.

---

### Why Strategy B (Pairs Trade) Works

The Gold/Silver ratio has strong **mean-reverting properties** over the medium term:
- The ratio has historically oscillated between ~40x and ~100x, reflecting relative industrial vs. monetary demand shifts.
- **Silver** has a dual nature: ~50% of demand is industrial (electronics, solar panels), making it more cyclically sensitive than gold.
- During risk-off, gold outperforms (ratio rises); during risk-on recovery, silver catches up (ratio falls).
- This creates exploitable spread dynamics.

**Cointegration caveat:** A proper pairs trade requires a statistically verified cointegrating relationship (e.g. Engle-Granger test). In a production system, this should be estimated rolling and the hedge ratio dynamically adjusted.

---

### Limitations

| Limitation | Impact | Mitigation |
|------------|--------|------------|
| yfinance futures = front-month only | Roll costs not modelled | Use Quandl/Bloomberg continuous adjusted data |
| Fixed slippage model | May underestimate costs in illiquid periods | Use volume-weighted or market impact model |
| No funding costs | Futures carry = convenience yield − storage | Add curve carry term |
| In-sample optimisation | Overfitting risk in sensitivity analysis | Use proper walk-forward or hold-out test set |
| No position limits / VaR constraint | Real funds have hard risk limits | Implement daily VaR cap |
| Correlations unstable | Gold/Silver can de-correlate in stress | Regime-conditional sizing |

---

### Key Takeaways

1. **Trend following on gold has historically delivered positive Sharpe ratios** due to persistent macro-driven momentum.
2. **Volatility scaling is critical** — unscaled trend strategies suffer large drawdowns in volatile periods.
3. **The combined portfolio diversifies** across two uncorrelated signal types (trend + mean reversion), improving risk-adjusted returns.
4. **Transaction costs are material** — the gross-to-net gap represents a real drag; reducing trade frequency (wider MA windows) can help.
5. **The Gold/Silver pairs trade is highly regime-dependent** — it performs best in stable, low-volatility environments where the ratio reverts predictably.
